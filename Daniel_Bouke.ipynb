# Getting information from tensors

* Shape
* Rank
* Axis or dimension
* Size

```python
#Create a rank 4 tensor(4 Dimensions)
rank_4_tensor=tf.zeros(shape=(2,3,4,5),dtype=tf.int32)
rank_4_tensor
```

```python
tf.shape(rank_4_tensor),rank_4_tensor.ndim,tf.size(rank_4_tensor),rank_4_tensor.dtype
```

```python
# Get various attributes of our tensor
print("Datatype of even element :",rank_4_tensor.dtype,"\n")
print("Number of dimennsions (rank)",rank_4_tensor.ndim,"\n")
print("Shape of tensor",rank_4_tensor.shape,"\n")
print("Element along of 0 axis ",rank_4_tensor[0],"\n")
print("Element along of last axis :",rank_4_tensor[-1],"\n")
print("Total number of element in our tensor :",tf.size(rank_4_tensor),"\n")
```

```python
# Get the first 2 element of each diagram
rank_4_tensor[:2,:2,:2,:2]
```

```python
rank_4_tensor[:1,:1,:1,:]
```
